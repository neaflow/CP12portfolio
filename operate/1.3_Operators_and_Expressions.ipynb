# 1.3 — Operators & Expressions

Variables are only useful once you can *do* something with them. Today's about the operators that combine and compare values — the arithmetic you'd expect, plus a couple of Python-specific ones worth knowing early.

## The Idea

The arithmetic operators: `+` `-` `*` `/` (regular division, always gives a `float`), `//` (floor division, drops the remainder), `%` (modulo, gives *just* the remainder), and `**` (exponent).

```python
17 // 5   # 3   -> how many times 5 fits into 17
17 % 5    # 2   -> what's left over
2 ** 3    # 8   -> 2 to the power of 3
```

Comparison operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) compare two values and give back a `bool` — these are what you'll feed into conditionals starting next lesson.

Python follows the same order of operations you learned in math class (exponents, then `*`/`/`/`//`/`%`, then `+`/`-`, left to right) — use parentheses `( )` whenever you want to force a different order.

````python
# A worked example — Zara deals damage to the monster

attack_power = 18
monster_defense = 5
damage_dealt = attack_power - monster_defense

monster_health = 40
monster_health = monster_health - damage_dealt

print("Damage dealt:", damage_dealt)
print("Monster health remaining:", monster_health)
print("Is the monster defeated?", monster_health <= 0)
````

## Guided Practice 2

````python
# Fill in the two blank values, then run — the operators are already correct

total_potions = 17
people_sharing = 2

potions_each = total_potions // people_sharing      # floor division — whole potions each
potions_leftover = total_potions % people_sharing   # modulo — what's left over

print("Potions each:", potions_each)
print("Potions leftover:", potions_leftover)
````
