## Quick Recall: Variables &amp; Data Types

Four types you'll use constantly: `str` (quoted text), `int` (whole number), `float` (decimal), `bool` (`True`/`False`, capitalized, no quotes). Python infers the type from the value — check any variable's type with `type()`.

```python
# Example
character_name = "Zara"
health = 100
damage_multiplier = 1.5
is_alive = True
print(character_name, health, damage_multiplier, is_alive)
```

## Quick Recall: Input/Output &amp; f-strings

`input("prompt")` always returns a `str` — wrap it in `int()`/`float()` for numbers. f-strings drop variables into text: `f"Hello, {name}!"`.

```python
# Example
player_name = input("Name? ")
gold = int(input("Starting gold? "))
print(f"Welcome, {player_name} — you begin with {gold} gold.")
```

## Quick Recall: Operators &amp; Expressions

`+ - * /` as expected; `//` floor-divides (drops the remainder), `%` gives just the remainder, `**` is exponent. Comparisons (`== != < > <= >=`) give back a `bool`.

```python
# Example
attack = 22
defense = 7
damage_dealt = attack - defense
print(damage_dealt, damage_dealt > 10)
```

## Quick Recall: Conditionals

`if` / `elif` / `else` runs the first `True` branch and skips the rest. Combine conditions with `and` / `or` / `not`. Remember: `=` assigns, `==` compares.

```python
# Example
monster_health = 12
if monster_health <= 0:
    print("Defeated.")
elif monster_health < 20:
    print("Nearly there.")
else:
    print("Still tough.")
```
