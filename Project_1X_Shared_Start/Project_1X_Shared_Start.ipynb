## A Note on Bad Input

You haven't been taught `try`/`except` directly, but you've seen enough of Python's error behavior to reason about it. Two viable approaches:

```python
try:
    gold = int(input("How much gold? "))
except ValueError:
    gold = 0
    print("That wasn't a number — defaulting to 0.")
```

or, without `try`/`except`, a validation loop using `.isdigit()`:

```python
raw = input("How much gold? ")
while not raw.isdigit():
    raw = input("Please enter a number: ")
gold = int(raw)
```

Either is acceptable — pick whichever makes more sense to you, or look up a third approach if you find one you understand well enough to explain.
